```python
import matplotlib.pyplot as plt
import networkx as nx


def draw_signed_graph(num_vertices, pos_edges, neg_edges):
    """Draws a signed graph with solid lines for positive edges and dashed lines for negative edges.

    Parameters:
    - num_vertices (int): Total number of vertices (labeled 1 to num_vertices).
    - pos_edges (list of tuples): Set of positive edges, e.g., [(1, 2), (2, 3)].
    - neg_edges (list of tuples): Set of negative edges, e.g., [(1, 3), (3, 4)].
    """
    G = nx.Graph()

    # Add vertices (labeled 1 to num_vertices)
    nodes = list(range(1, num_vertices + 1))
    G.add_nodes_from(nodes)

    # Compute node positions using circular layout for clear readability
    pos = nx.circular_layout(G)

    plt.figure(figsize=(8, 6))

    # Draw nodes and labels
    nx.draw_networkx_nodes(
        G, pos, node_color="skyblue", node_size=700, edgecolors="black"
    )
    nx.draw_networkx_labels(G, pos, font_weight="bold", font_size=12)

    # Draw positive edges (solid lines)
    if pos_edges:
        nx.draw_networkx_edges(
            G,
            pos,
            edgelist=pos_edges,
            style="solid",
            edge_color="green",
            width=2,
            label="Positive (+)",
        )

    # Draw negative edges (dashed lines)
    if neg_edges:
        nx.draw_networkx_edges(
            G,
            pos,
            edgelist=neg_edges,
            style="dashed",
            edge_color="red",
            width=2,
            label="Negative (-)",
        )

    plt.title("Signed Graph Visualization", fontsize=14, fontweight="bold")
    plt.legend(loc="upper right")
    plt.axis("off")
    plt.tight_layout()
    plt.show()


# Example Usage:
if __name__ == "__main__":
    num_vertices = 5
    positive_edges = [(1, 2), (2, 3), (4, 5)]
    negative_edges = [(1, 3), (3, 4), (2, 5)]

    draw_signed_graph(num_vertices, positive_edges, negative_edges)
```
